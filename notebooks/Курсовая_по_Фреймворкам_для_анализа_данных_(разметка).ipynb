# Датасет с изображениями

Ссылка на оригинальный датасет: https://www.kaggle.com/datasets/pierreayfri/minecraft-mobs

В исходном датасете есть таблица frames.csv, в ней содержатся данные о погоде, месте съемки моба, его типе, времени. Эта таблица не заполнялась для новых объектов.

Замечание по поводу классов: так как в оригинальном датасете есть более 87 мобов, было принято решение использовать другую систему классов - по поведению мобов (враждебность, на странице датасета явно указан соответствующий тип поведения для каждого моба, но там есть ряд ошибок, они учтены при преобразовании) Ниже приведен код, преобразующий классы из оригинального датасета в выбранные. (Passive - 0, Hostile - 1, Neutral - 2, Boss/Other - 3)

```python
def transform_cls(cls_id: int):
  if (cls_id == 83):
    return 0
  if (cls_id == 41): # обмен местами железного и медного големов
    return 3
  if (cls_id < 37):
    return 1
  if (cls_id < 48):
    return 2
  if (cls_id < 83):
    return 0
  return 3

path = ""
df = pd.read_csv(path)
df["class_id"] = df["class_id"].apply(transform_cls)
df.to_csv(path, sep=",")
```

Правила разметки
1. Если опозноваемый объект частично закрыт другим, всё ещё определяем его как валидный.
2. Если опозноваемы объект находится на изображении не полностью, определяем его как валидный.
